```python
df = spark.read.table("leona.purchaseorder.purchaseorderitem")
display(df)
```

```python
missing_counts = df.select([((df[c].isNull()).cast("int")).alias(c) for c in df.columns]).groupBy().sum().collect()[0]
missing_dict = {col: missing_counts[i] for i, col in enumerate(df.columns)}

plt.figure(figsize=(10, 6))
plt.bar(missing_dict.keys(), missing_dict.values())
plt.ylabel('Number of Missing Values')
plt.xlabel('Columns')
plt.title('Missing Values per Column')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
