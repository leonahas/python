```python
df = spark.read.table("leona.purchaseorder.purchaseorderitem")
display(df)
```

```python
missing_counts = df.select([((df[c].isNull()).cast("int")).alias(c) for c in df.columns]).groupBy().sum().collect()[0]
missing_dict = {col: missing_counts[i] for i, col in enumerate(df.columns)}

plt.figure(figsize=(10, 6))
plt.bar(missing_dict.keys(), missing_dict.values())
plt.ylabel('Number of Missing Values')
plt.xlabel('Columns')
plt.title('Missing Values per Column')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
from pyspark.sql.functions import col, sum as spark_sum, isnan, when
from pyspark.sql.types import DoubleType, FloatType

missing_df = df.select([
    spark_sum(
        when(col(c).isNull() | (isnan(col(c)) if isinstance(df.schema[c].dataType, (DoubleType, FloatType)) else False), 1)
        .otherwise(0)
    ).alias(c)
    for c in df.columns
])

missing_df = missing_df.transpose()
display(missing_df)
```
